```python
teams = df[df["position"] == "team"].copy()
teams["date"] = pd.to_datetime(teams["date"])
print(teams.groupby(teams["date"].dt.to_period("M"))["gameid"].nunique())
print(teams[teams["year"] == 2026][["date", "league", "split"]].drop_duplicates().head())
```

```python
blue = teams[teams["side"] == "Blue"][
    ["gameid", "date", "league", "split", "playoffs", "patch", "teamname", "result"]
].rename(columns={"teamname": "blue_team", "result": "blue_win"})

red = teams[teams["side"] == "Red"][["gameid", "teamname"]].rename(
    columns={"teamname": "red_team"}
)

games = blue.merge(red, on="gameid").sort_values("date").reset_index(drop=True)
print(games.shape)
games.head()
```
